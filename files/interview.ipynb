## 第一题：量化回测框架问题分析

一个量化团队考虑开发一个预测10天后的股价回报的模型，考虑下边这个回测框架：

- 把历史数据按日随机分成不重合的两类，第一类是训练数据占比80%，第二类是测试数据占比20%，每个交易日对每个股票计算其十天后的收盘价，除以当前收盘价，再减去1，得到回报数据，这个是我们的Y；同时，清洗我们的数据源，并计算出600个特征值，其中有100个是量价特征值，500个是基本面特征值，量价特征值都是使用每天收盘价和交易量计算的，基本面特征值都是使用公司最近一期的财报数据计算的。

- 在训练集上用10-fold cross validation训练出来一个模型，得到所有参数，用这些参数和模型在测试数据上验算并使用 IC作为评价标准，如果结果不佳，尝试修改特征值的定义，使用更高阶的函数形式，不断迭代，直到得到自己满意的答案。
